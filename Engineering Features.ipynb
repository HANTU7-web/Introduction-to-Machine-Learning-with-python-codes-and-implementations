```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt


demo_df = pd.DataFrame({'Categorical Feature': ['socks', 'fox', 'socks', 'box'],
                        'Integer feature': [0,1,2,1]})
display(demo_df)
```

```python
display(pd.get_dummies(demo_df))
```

```python
## Recent updates to the pandas library changed the default output of pd.get_dummies() to booleans (True/False) instead of integers (1/0).To force the function to return 1 and 0, simply add the dtype=int parameter to your code:
# New Code with dtype=int or dtype=float

display(pd.get_dummies((demo_df),dtype= float))
```
